# 3. Data Cleaing on NFHS-5 Factsheet

```python
columns_to_keep = [
        'States/UTs',
        'Area',
        'Number of Households surveyed',
        'Population living in households with electricity (%)',
        'Households using clean fuel for cooking3 (%)',
        'Women (age 15-49) who are literate4 (%)',
        'Men (age 15-49) who are literate4 (%)',
        'Women (age 15-49)  with 10 or more years of schooling (%)',
        'Men (age 15-49)  with 10 or more years of schooling (%)',
        'Women (age 15-49)  who have ever used the internet (%)',
        'Men (age 15-49)  who have ever used the internet (%)',
        'Women (age 15-49 years) having a bank or savings account that they themselves use (%)',
        'Women (age 15-49 years) having a mobile phone that they themselves use (%)',
        'Currently married women (age 15-49 years) who usually participate in three household decisions25 (%)'
    ]
df_datafile = df_datafile[columns_to_keep]
```

```python
df_datafile.head()
```

### 1. Rename Columns

```python
num_cols = df_datafile.columns[2:]
for c in num_cols:
    df_datafile[c] = pd.to_numeric(df_datafile[c].astype(str).str.strip().replace({'*': np.nan, '-': np.nan, '': np.nan}), errors='coerce')
```

```python
columns_to_keep = [
        'States/UTs',
        'Households using clean fuel for cooking3 (%)',

        'Women (age 15-49) who are literate4 (%)',
        'Men (age 15-49) who are literate4 (%)',

        'Women (age 15-49)  with 10 or more years of schooling (%)',
        'Men (age 15-49)  with 10 or more years of schooling (%)',

        'Women (age 15-49)  who have ever used the internet (%)',
        'Men (age 15-49)  who have ever used the internet (%)',

        'Women (age 15-49 years) having a bank or savings account that they themselves use (%)',
        'Women (age 15-49 years) having a mobile phone that they themselves use (%)',

        'Population living in households with electricity (%)',
        'Currently married women (age 15-49 years) who usually participate in three household decisions25 (%)',
        'Number of Households surveyed',
    ]

to_rename = [
    "state_ut",
    "households_clean_fuel_pct",
    
    "female_literacy_pct",
    "male_literacy_pct",

    "female_schooling_pct",
    "male_schooling_pct",
    
    "Women_Internet_pct", 
    "Men_Internet_pct", 
    
    "women_bank_account_pct", 
    "Women_Mobile_Ownership_pct", 
    
    "Electricity_Access",
    "women_household_decisions_pct",
    "households_surveyed"
]
df_data = df_datafile.copy()

df_data = df_data.rename(
    columns=dict(zip(columns_to_keep, to_rename))
)
```

```python
df_data.head()
```

```python
col = [
    "households_clean_fuel_pct",
    "female_literacy_pct",
    "male_literacy_pct",
    "female_schooling_pct",
    "male_schooling_pct",
    "Women_Internet_pct", 
    "Men_Internet_pct", 
    "women_bank_account_pct", 
    "Women_Mobile_Ownership_pct",
    "women_household_decisions_pct",
]
for i in col:
    df_data[i].astype("float64")
```

```python
float_cols = df_data.select_dtypes(include=['float64']).columns

# 2. Apply the absolute value function to just those columns
df_data[float_cols] = df_data[float_cols].abs()
```

### 2. Pivoting & Base Table Construction

```python
df_total = df_data[df_datafile['Area'] == 'Total'].copy()
df_rural = df_data[df_datafile['Area'] == 'Rural'].copy()
df_urban = df_data[df_datafile['Area'] == 'Urban'].copy()
```

```python
# Remove area column
df_total = df_total.drop(columns='Area')
df_rural = df_rural.drop(columns='Area')
df_urban = df_urban.drop(columns='Area')
```

```python
# Add area suffix
df_total = df_total.rename(
    columns=lambda x: x if x == 'state_ut' else f'{x}_total'
)

df_rural = df_rural.rename(
    columns=lambda x: x if x == 'state_ut' else f'{x}_rural'
)

df_urban = df_urban.rename(
    columns=lambda x: x if x == 'state_ut' else f'{x}_urban'
)
```

```python
df_master = (
    df_total
    .merge(df_rural, on='state_ut', how='inner')
    .merge(df_urban, on='state_ut', how='inner')
)
```
